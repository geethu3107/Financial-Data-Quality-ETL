```python
import pandas as pd

# Load source data
file_path = r"C:\Users\Lenovo\Downloads\complaints-2026-09-23_12_47.csv"
df = pd.read_csv(file_path)

# Basic profile
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
print(df.head())

# -----------------------------
# Completeness Check
# -----------------------------

null_summary = pd.DataFrame({
    "Column": df.columns,
    "Null_Count": df.isna().sum().values,
    "Null_Percentage": (df.isna().mean() * 100).round(2).values
})

print("\nMissing Value Summary:")
print(null_summary)

# -----------------------------
# Complaint ID Uniqueness
# -----------------------------

total_records = len(df)
unique_ids = df["Complaint ID"].nunique()
duplicate_ids = df["Complaint ID"].duplicated().sum()

print("\nRecord Validation:")
print("Total records:", total_records)
print("Unique Complaint IDs:", unique_ids)
print("Duplicate Complaint IDs:", duplicate_ids)

# -----------------------------
# Categorical Profiling
# -----------------------------

categorical_columns = [
    "Product",
    "Sub-product",
    "Issue",
    "Sub-issue",
    "Company",
    "State",
    "Submitted via",
    "Company response to consumer",
    "Timely response?"
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(10))

# -----------------------------
# Save Profile
# -----------------------------

profile = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Total_Records": len(df),
    "Non_Null": df.notna().sum().values,
    "Null_Count": df.isna().sum().values,
    "Null_Percentage": (df.isna().mean() * 100).round(2).values,
    "Distinct_Values": df.nunique(dropna=True).values
})

profile.to_csv(
    "source_profile.csv",
    index=False
)

print("\nProfile saved successfully.")
```
