```python
n=float(input("Enter the consumed units:"))
def bill():
    if n<50:
       return 100
    elif 50<n<300:
       return n*4.5
    elif n>300:
        return 1000+((n-300)*7)
print(bill())
           
```

```python
n=float(input("Enter the salary:"))
def tax():
    if n>150000:
        return 0.18*n
    elif 100000<n<= 150000:
        return 0.15*n
    elif 0<n<= 100000:
        return 0.1*n
print(tax())
         
    
```
